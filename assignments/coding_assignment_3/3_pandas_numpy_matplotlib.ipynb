#### 판다스의 데이터 구조
- **Series**: 1차원 배열의 값에 대응하는 인덱스를 부여할 수 있는 구조
- **DataFrame**: 2차원 리스트를 매개변수로 전달
- **Panel**: 3차원 데이터를 다룰 때 사용. 잘 쓰이지 않음

```python
# 시리즈 클래스: 1차원 배열의 값(values)에 각 값에 대응하는 인덱스를 부여할 수 있는 구조를 갖고 있다.

sr = pd.Series([17000, 18000, 1000, 5000], index=["피자", "치킨", "콜라", "맥주"])
print('시리즈 출력: ')
print('-'*15)
print(sr)
```

```python
print("시리즈의 값: {}".format(sr.values))
print("시리즈의 인덱스: {}".format(sr.index))
```

```python
# 데이터 프레임: 2차원 리스트를 매개변수로 전달

values = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
index = ['one', 'two', 'three']
columns = ['A', 'B', 'C']

df = pd.DataFrame(values, index=index, columns=columns)

print('데이터프레임 출력: ')
print('-'*18)
print(df)
```

```python
# 데이터프레임 인덱스, 값, 열 출력
print('데이터프레임의 인덱스 : {}'.format(df.index))
print('데이터프레임의 열이름 : {}'.format(df.columns))
print('데이터프레임의 값 : {}'.format(df.values))
print('-'*18)
print(df.values)
```

```python
# 데이터프레임의 생성
  # 리스트, 시리즈, 딕셔너리, Numpy의 ndarrays, 또 다른 데이터프레임으로부터 생성 가능

# 리스트로 데이터프레임 생성하기
data = [
    ['1000', 'Steve', 90.72],
    ['1001', 'James', 78.09],
    ['1002', 'Doyeon', 98.43],
    ['1003', 'Jane', 64.19],
    ['1004', 'Pilwoong', 81.30],
    ['1005', 'Tony', 99.14]
]

df = pd.DataFrame(data)
print(df)
print(type(df))

```

###NumPy (넘파이)
수치 데이터를 다루는 파이썬 패키지

- ndarray: 다차원 행렬자료구조. 넘파이의 핵심

```python
# 넘파이 설치
!pip install numpy
```

```python
# 넘파이 버전 확인

import numpy as np
np.__version__
```

```python
# np.array(): 리스트, 튜플, 배열로부터 ndarray 생성

# 1차원 배열
vec = np.array([1, 2, 3, 4, 5])
print(vec)
```

```python
# 2차원 배열

mat = np.array([[10, 20, 30], [60, 70, 80]])
print(mat)
```

```python
# 위 두 배열의 타입 확인

print('vec의 타입: ', type(vec))
print('mat의 타입: ', type(mat))
```

```python
# 배열 축의 개수와 크기

print('vec의 축의 개수; ', vec.ndim)
print('vec의 크기(shape): ', vec.shape)

print('mat의 축의 개수: ', mat.ndim)
print('mat의 크기(shape): ', mat.shape)
```

##### ndarray의 초기화

```python
# np.zeros: 배열의 모든 원소에 0 삽입

zero_mat = np.zeros((2, 3)) # 모든 값이 0인 2x3 배열 생성
print(zero_mat)
```

```python
# np.ones: 배열의 모든 원소에 1 삽입

one_mat = np.ones((2, 3)) # 모든 값이 1인 2X3 배열 생성
print(one_mat)
```

```python
# np.full: 배열에 사용자가 지정한 값 삽입

same_value_mat = np.full((2,2), 7)  # 모든 값이 7인 2X2 배열 생성
print(same_value_mat)
```

```python
# np.eye: 대각선은 1이고, 나머지는 0인 2차원 배열 생성
eye_mat = np.eye(3)
print(eye_mat)
```

```python
# np.random.random: 임의의 값을 가지는 배열 생성
random_mat = np.random.random((2, 2))
print(random_mat)
```

```python
# np.arrange(n): 0부터 n-1까지의 값을 가지는 배열 생성
range_vec = np.arange(10)
print(range_vec)
```

```python
# np.arange(i, j, k): i부터 j-1까지 k씩 증가하는 배열 생성
n=2
range_n_step_vec = np.arange(1, 10, n)
print(range_n_step_vec)
```

```python
# np.reshape(): 내부 데이터는 변경하지 않으면서 배열의 구조 바꾸기

orig_mat = np.arange(30)  # 0부터 29까지의 값을 가지는 배열 생성
print(orig_mat)

reshape_mat = np.array(orig_mat).reshape((5, 6))  # orig_mat의 배열 구조를 (5, 6) 행렬로 바꾸기
print(reshape_mat)
```

```python
# numpy slicing

mat = np.array([[1, 2, 3], [4, 5, 6]]) # 배열 생성
print(mat)

slicing_mat = mat[0, :]  # 첫번째 행 출력
print(slicing_mat)

slicing_mat = mat[:, 1]  # 두번째 열 출력
print(slicing_mat)
```

```python
# 정수 인덱싱

mat = np.array([[1, 2], [4, 5], [7, 8]])  # 배열 생성
print(mat)

print(mat[1, 0]) # mat 행렬의 1행 0열 원소 (행이 0부터 시작하므로 1행 0열 원소는 4)

indexing_mat = mat[[2, 1], [0, 1]]  # 2행 0열, 1행 1열 두개 원소
print(indexing_mat)
```

```python
# 넘파이 연산

x = np.array([1, 2, 3])
print('x_array', x)
y = np.array([4, 5, 6])
print('y_array', y)
print('\n')

result = x + y
print('result_plus', result)

result1 = x - y
print('result_minus', result1)

result2 = result1 * x
print('result_multiply', result2)

result3 = result2 /  x
print('result_divide', result3)

print('\n')

mat1 = np.array([[1, 2], [3, 4]])
print('mat1', mat1)

mat2 = np.array([[5, 6], [7, 8]])
print('mat2', mat2)

mat3 = np.dot(mat1, mat2)
print('mat_dot', mat3)
```

```python
# pyplot(맷플롯립의 주요 모듈) 임포트
import matplotlib.pyplot as plt
```

```python
# 라인 플롯 그리기
plt.title('test')
plt.plot([1, 2, 3, 4], [2, 4, 8, 6])
plt.show()
```

```python
# 축 레이블 삽입
plt.title('test')
plt.plot([1, 2, 3, 4], [2, 4, 8, 6])
plt.xlabel('hours')
plt.ylabel('score')
plt.show()
```

```python
# 라인 추가와 범례 삽입

plt.title('students')
plt.plot([1, 2, 3, 4], [2, 4, 8, 6])
plt.plot([1.5, 2.5, 3.5, 4.5], [3, 5, 8, 10])  # 라인 새로 추가
plt.xlabel('hours')
plt.ylabel('score')
plt.legend(['A student', 'B student'])  # 범례 삽입
plt.show()
```
